#### 4. How to find the memory size of any array (★☆☆)

````python
#Each element size
a = arr.itemsize
print("Each element size:", a, "bytes")

#Data size
b = arr.size * a
print("Data size: ", b, "bytes")

#Total size: Metadata + data
c = arr.__sizeof__()
print("Total size: ",c,"bytes")
````

#### 20. Consider a (6,7,8) shape array, what is the index (x,y,z) of the 100th element? (★☆☆)

````python
index = 100-1

x = index//(7*8)
index -= 7*8*x

y = index//8
index -= 8*y

z = index
index -= z

print(x, y, z)


#Test
arr = np.zeros(6*7*8, dtype="int32")
arr[99] = 1
arr = arr.reshape(6, 7, 8)
print(arr)
````

#### 21. Create a checkerboard 8x8 matrix using the tile function (★☆☆)

#### 22. Normalize a 5x5 random matrix (★☆☆)

````python
arr = rng.uniform(0, 30, (5,5))

print(arr)
print()

#Simple min-max
maxim = arr.max()
minim = arr.min()
arr2 = (arr - minim)/(maxim - minim)
print("Min-Max")
print(arr2)
print()

#Per-Column standarization
mu = arr.mean(axis=0)
sigma = arr.std(axis=0, ddof=1)

arr2 = (arr-mu)/sigma

print("Standarization")
print(arr2)
print()

````

#### 35. How to compute ((A+B)*(-A/2)) in place (without copy)? (★★☆)

````python
A = rng.uniform(0, 10, 10)
B = rng.uniform(0, 10, 10)

A += B
A *= np.divide(np.multiply(-1.0, np.subtract(A, B, out=A), out=A), 2.0)

print(A)
````
